## 🌦️ 1. Season Finder for a Travel App

### Goal:
    Tell travellers which season it is for a given month.

### Steps:

    Ask the user for a month number (1-12).
    
    December, January, February → "Winter".
    
    March, April, May → "Spring".
    
    June, July, August → "Summer".
    
    September, October, November → "Autumn".
    
    Any other number → "Invalid month".

### Example:

    Enter month number: 7
    Summer

```python
month = int(input("Enter month number : "))
if month >= 1 and month <= 12:
    if month == 12 or (month <= 2):
        print("Winter")
    elif month <= 5:
        print("Spring")
    elif month <= 8:
        print("Summer")
    else:
        print("Autumn")
else:
    print("Invalid month")
```

## ⚖️ 2. BMI Health Checker

### Goal:
    Help a fitness app tell users which weight category they belong to.

### Steps:

    Ask for weight in kg and height in metres.
    
    Calculate BMI = weight / (height * height).
    
    BMI below 18.5 → "Underweight".
    
    18.5 to 24.9 → "Normal".
    
    25 to 29.9 → "Overweight".
    
    30 or more → "Obese".
    
    Print the BMI rounded to 2 decimals.

### Example:

    Enter weight: 70
    Enter height: 1.75
    BMI = 22.86 → Normal

```python
weight = float(input("Enter weight : "))
height = float(input("Enter height : "))
bmi = weight / height ** 2
weight_cat = ""
if bmi < 18.5:
    weight_cat = "Underweight"
elif bmi <= 24.9:
    weight_cat = "Normal"
elif bmi <= 29.9:
    weight_cat = "Overweight"
else:
    weight_cat = "Obese"
print(f"You are {weight_cat} and your bmi is {round(bmi, 2)}")
```

## 🌡️ 8. Temperature Unit Converter

### Goal:
    Build a small converter for a weather app.

### Steps:

    Show a menu: 1 = Celsius to Fahrenheit, 2 = Fahrenheit to Celsius.
    
    Ask for the choice and then the temperature.
    
    Choice 1 → F = C * 9/5 + 32.
    
    Choice 2 → C = (F - 32) * 5/9.
    
    Any other choice → "Invalid choice".

### Example:

    Enter choice: 1
    Enter temperature: 100
    212.0 F

```python
print("Press 1 if you want to convert temperature from Celsius to Fahrenheit")
print("Press 2 if you want to convert temperature from Fahrenheit to Celsius")
user_choice = int(input("Enter your choice : "))
temp = 0
if user_choice == 1:
    temp = float(input("Enter temperature : "))
    temp = temp * 9/5 + 32
    print(f"Temperature in Fahrenheit : {temp}F")
elif user_choice == 2:
    temp = float(input("Enter temperature : "))
    temp = (temp - 32) * 5/9
    print(f"Temperature in Celsius : {temp}C")
else:
    print("Invalid choice")
```

## 🛒 9. Cheapest Price Finder

### Goal:
    Help a shopper find the lowest price of the same product from three shops.

### Steps:

    Ask for the price in Shop 1, Shop 2 and Shop 3.
    
    Use nested if statements to find the smallest price (do not use min()).
    
    Print the lowest price and the shop number.

### Example:

    Enter prices: 1450, 1200, 1380
    Lowest price: 1200 (Shop 2)

```python
shop_one_price = int(input("Enter shop one price : "))
shop_two_price = int(input("Enter shop two price : "))
shop_three_price = int(input("Enter shop three price : "))

if shop_one_price < shop_two_price:
    if shop_one_price < shop_three_price:
        print(f"Minimum price : {shop_one_price}")
    else:
        print(f"Minimum price : {shop_three_price}")

elif shop_two_price < shop_one_price:
    if shop_two_price < shop_three_price:
        print(f"Minimum price : {shop_two_price}")
    else:
        print(f"Minimum price : {shop_three_price}")

else:
    print(f"All shops price are equal : {shop_one_price}")
```

## 🅿️ 10. Parking Fee System

### Goal:
    Calculate the parking fee of a mall for different vehicles.

### Steps:

    Ask for the vehicle type (car or bike).
    
    Then ask for the number of hours parked.
    
    Car: Rs 100 for up to 2 hours, plus Rs 50 for each extra hour.
    
    Bike: Rs 50 for up to 2 hours, plus Rs 20 for each extra hour.
    
    Any other vehicle → "Vehicle not allowed".

### Example:

    Enter vehicle: car
    Enter hours: 5
    Parking fee = Rs 250

```python
vehicle_type = input("Enter vehicle type").lower()
parking_fee = 0
hours = int(input("Enter hours : "))
if vehicle_type == "car":
    if hours <= 2:
        parking_fee += 100
        print(f"Parking fee : {parking_fee}")
    elif hours > 2:
        parking_fee = 100 + (hours - 2) * 50
        print(f"Parking fee : {parking_fee}")
elif vehicle_type == "bike":
    if hours <= 2:
        parking_fee += 50
        print(f"Parking fee : {parking_fee}")
    elif hours > 2:
        parking_fee = 50 + (hours - 2) * 20
        print(f"Parking fee : {parking_fee}")
else:
    print("Vehicle not allowed")
```

## 🚕 11. Taxi Fare Calculator

### Goal:
    Calculate the fare of a ride-booking app.

### Steps:

    Ask for the ride type (economy or luxury).
    
    Ask for the distance in km.
    
    Economy: Rs 200 for up to 5 km, then Rs 30 per extra km.
    
    Luxury: Rs 400 for up to 5 km, then Rs 60 per extra km.
    
    Print the final fare.

### Example:

    Enter ride type: economy
    Enter distance: 8
    Fare = Rs 290

```python
ride_type = input("Enter ride type").lower()
fare = 0
distance = int(input("Enter distance : "))
if ride_type == "economy":
    if distance <= 5:
        fare += 200
        print(f"Fare : {fare}")
    else:
        fare = 200 + (distance - 5) * 30
        print(f"Fare : {fare}")
elif ride_type == "luxury":
    if distance <= 5:
        fare += 400
        print(f"Fare : {fare}")
    else:
        fare = 400 + (distance - 5) * 60
        print(f"Fare : {fare}")
else:
    print("Vehicle not allowed")
```

## ✊ 17. Rock, Paper, Scissors Game

### Goal:
    Decide the winner of a two-player game.

### Steps:

    Ask Player 1 and Player 2 for rock, paper or scissors.
    
    Rock beats scissors, scissors beats paper, paper beats rock.
    
    Use logical operators (and / or) to print the winner.
    
    If both choose the same → "Draw".

### Example:

    Player 1: rock
    Player 2: scissors
    Player 1 wins

```python
pla_one = input("Please select rock, paper or scissors: ").lower()
pla_two = input("Please select rock, paper or scissors : ").lower()
if pla_one == "rock" and pla_two == "paper":
    print("Player two wins")
elif pla_one == "rock" and pla_two == "scissors":
    print("Player one wins")
elif pla_one == "scissors" and pla_two == "paper":
    print("Player one wins")
elif pla_one == "scissors" and pla_two == "rock":
    print("Player two wins")
elif pla_one == "paper" and pla_two == "rock":
    print("Player one wins")
elif pla_one == "paper" and pla_two == "scissors":
    print("Player two wins")
else:
    print("Draw")
```

## 📐 20. Roof Truss Angle Check

### Goal:
    Help a civil engineer check whether three angles can form a triangle.

### Steps:

    Ask for three angles.
    
    Valid only if every angle is greater than 0 AND the sum of the three angles is exactly 180.
    
    Print "Valid triangle" or "Invalid triangle".

### Example:

    Enter angles: 60, 60, 60
    Valid triangle

```python
angles = []
all_angles_greater_than_zero = True
for i in range(3):
    angles.append(int(input("Enter angle : ")))
for angle in angles:
    if angle <= 0:
        all_angles_greater_than_zero = False
        break
if all_angles_greater_than_zero and sum(angles) == 180:
    print("Valid triangle")
else:
    print("Not valid triangle")
```

## 🎟️ 35. Lucky Ticket Digit Sum

### Goal:
    A raffle adds up the digits of every ticket number.

### Steps:

    Ask for a ticket number.
    
    Use a while loop with % and // to pick out each digit.
    
    Add the digits together.
    
    Print the digit sum.

### Example:

    Ticket: 4582
    Digit sum = 19

```python
ticket = int(input("Enter ticket no : "))
temp_ticket = ticket
digits_len = len(str(ticket))
digit = 0
digit_sum = 0
start = True
while start:
    no_to_be_divided = 10 ** (digits_len - 1)
    if no_to_be_divided >= 10:
        digit = temp_ticket // no_to_be_divided
        digit_sum = digit + digit_sum
        temp_ticket = temp_ticket - digit * no_to_be_divided
    else:
        digit = temp_ticket
        digit_sum = digit + digit_sum
        start = False
    digits_len -= 1
print(f"Ticket : {ticket}")
print(f"Digit sum : {digit_sum}")        
```

## 💵 45. ATM Cash Notes Counter

### Goal:
    An ATM splits a withdrawal into notes of 5000, 1000 and 500.

### Steps:

    Create notes = [5000, 1000, 500] and ask for the amount (a multiple of 500).
    
    Loop through the notes.
    
    Number of notes = amount // note, then amount = amount % note.
    
    Print a line only if the number of notes is greater than 0.

### Example:

    Amount: 8500
    5000 x 1
    1000 x 3
    500 x 1

```python
notes = [5000, 1000, 500]
amount = int(input("Enter amount which is multiple of 500: "))
five_thousand_notes_partition = 0
thousand_notes_partition = 0
five_hundred_notes_partition = 1
note_splitting_complete = False
remaining_amount = 0
if amount % 500 == 0:
    for note in notes:
        if note == 5000:
            if not note_splitting_complete and amount % 5000 == 0:
                five_thousand_notes_partition = amount / note
                note_splitting_complete = True
            else:
                five_thousand_notes_partition = amount // note
                remaining_amount = amount - (five_thousand_notes_partition * 5000)
                 
        elif note == 1000:
            if not note_splitting_complete and remaining_amount % 1000 == 0:
                thousand_notes_partition = remaining_amount / note
            else:
                if five_thousand_notes_partition == 0:
                    thousand_notes_partition = amount // note
                else:
                    thousand_notes_partition = remaining_amount // note
        else:
              if amount % 1000 == 0 or amount % 5000 == 0:
                  five_hundred_notes_partition = 0
    print(f"5000 x {five_thousand_notes_partition} = {five_thousand_notes_partition}")
    print(f"1000 x {thousand_notes_partition} = {thousand_notes_partition}")
    print(f"500 x {five_hundred_notes_partition} = {five_hundred_notes_partition}")
else:
    print("You have enter invalid amount")
```

## 🧱 47. Largest Square Tile

### Goal:
    A floor is 48 ft by 18 ft. Find the biggest square tile that fits without cutting.

### Steps:

    Ask for the length and the width.
    
    Use a while loop with Euclid's method: while b is not 0, set a, b = b, a % b.
    
    The answer is the final value of a (the GCD).
    
    Print the tile size.

### Example:

    Length: 48
    Width: 18
    Largest tile = 6 ft

```python
a = int(input("Enter the length of floor : "))
b = int(input("Enter the width of floor : "))
length = a
width = b
while b != 0:
    a, b = b, a % b
tile_size = a
print(f"Length : {length}")
print(f"Width : {width}")
print(f"Largest tile : {tile_size}")
```

## 🎫 49. Roll Number Order Check

### Goal:
    A teacher checks whether roll numbers were entered in ascending order.

### Steps:

    Create a list of roll numbers.
    
    Loop through the list and compare each number with the next one.
    
    Use a flag variable that becomes False if any number is out of order.
    
    Print "Sorted" or "Not sorted".

### Example:

    [101, 102, 105, 110] → Sorted
    [101, 105, 103] → Not sorted

```python
roll_nos = [101, 102, 105, 110]
is_sorted = True
for i in range(len(roll_nos)):
    for j in range(i + 1, len(roll_nos)):
        if roll_nos[i] > roll_nos[j]:
            is_sorted = False
if is_sorted:
    print("Sorted")
else:
    print("Not sorted")
```

## 🌡️ 50. Weekly Temperature Report

### Goal:
    A weather station wants to know how many days were warmer than average.

### Steps:

    Create temps = [30, 32, 28, 35, 31, 29, 36].
    
    First loop: add all the temperatures and calculate the average.
    
    Second loop: count the days above the average.
    
    Print the average (2 decimals) and the count.

### Example:

    Average = 31.57
    Days above average = 3

```python
temps = [30, 32, 28, 35, 31, 29, 36]
temp_sum = 0
days_above_avg = 0
for temp in temps:
    temp_sum += temp
avg_temp = round(temp_sum / len(temps), 2)
for temp in temps:
    if temp > avg_temp:
        days_above_avg += 1
print(f"Average : {avg_temp}")
print(f"Days above average : {days_above_avg}")
```
